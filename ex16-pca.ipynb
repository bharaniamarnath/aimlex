```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
X_train = ss.fit_transform(X_train)
```

```python
X_train = pd.DataFrame(X_train, columns=['Age', 'Gender', 'Ethnicity', 'ParentalEducation', 'StudyTimeWeekly',
       'Absences', 'Tutoring', 'ParentalSupport', 'Extracurricular', 'Sports',
       'Music', 'Volunteering', 'GPA'])
```

```python
X_train.head(10)
```

```python
X_train.describe().round(3)
```

```python
from sklearn.decomposition import PCA
pca1 = PCA()
```

```python
Xtpca1 = pca1.fit_transform(X_train)
```

```python
pca1.explained_variance_ratio_
```

```python
plt.plot(pca1.explained_variance_ratio_)
plt.xlabel("No of comps")
plt.ylabel("Exp var rat")
plt.show()
```

```python
pca2 = PCA(0.95)
Xtpca2 = pca2.fit_transform(X_train)
```

```python
pca2.explained_variance_ratio_
```

```python
plt.plot(pca2.explained_variance_ratio_)
plt.xlabel("No of comps")
plt.ylabel("Exp var rat")
plt.show()
```

```python
Xtpca1.shape
```

```python
pca3 = PCA(n_components=2)
Xtpca3 = pca3.fit_transform(X_train)
```

```python
xcmap = plt.cm.get_cmap("coolwarm")
```

```python
plt.figure()
xsplt = plt.scatter(Xtpca3[:,0], Xtpca3[:,1], c=y_train, cmap=xcmap)
plt.xlabel("xpca3 col0")
plt.ylabel("xpca3 col1")
plt.colorbar(xsplt, label="grade class")
plt.show()
```
