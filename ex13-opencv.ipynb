```python
import numpy as np
import pandas as pd
from glob import glob
import cv2
import matplotlib.pylab as plt
```

```python
dogs = glob("../Datasets/PetImages/Dog/*.jpg")
cats = glob("../Datasets/PetImages/Cat/*.jpg")
```

```python
imgplt = plt.imread(dogs[20])
imgcv2 = cv2.imread(dogs[20])
```

```python
fig, ax = plt.subplots(figsize=(10,10))
ax.imshow(imgplt)
ax.axis('off')
plt.show()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(8, 8))
ax[0].imshow(imgplt[:,:,0], cmap="Reds")
ax[1].imshow(imgplt[:,:,1], cmap="Greens")
ax[2].imshow(imgplt[:,:,2], cmap="Blues")
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(8, 8))
ax[0].imshow(imgplt)
ax[1].imshow(imgcv2)
plt.show()
```
