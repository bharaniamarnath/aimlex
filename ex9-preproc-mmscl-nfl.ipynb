```python
ds = ds[ds["player_id"].duplicated() != 1]
ds = ds.fillna(0)
ds = ds.drop(columns=['player_id','player_name','position','draft_year','birth_year','college','team','conference','division'])
ds
```

```python
from sklearn.preprocessing import StandardScaler
X1 = ds.iloc[:,:412]
X2 = ds.iloc[:,:4120]
sscaler = StandardScaler()
X1 = sscaler.fit_transform(X1)
X1
```

```python
dscols = ds.columns
X1 = pd.DataFrame(X1, columns=dscols)
X1
```

```python
sns.heatmap(X1[['season','games_played_career','games_played_season','games_missed','seasons_played']].corr(),annot=True)
```
